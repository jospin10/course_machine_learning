## 2 Constructeur d’initialisation des tableaux: zeros, ones, random, eye

```python
# Tableau contenant que des zeros de taille 10
my_arr0 =  np.zeros(10, dtype = np.int8)
print(my_arr0)
print(my_arr0.shape)

# Tableau de dim 2D (4,5)
my_arr2d = np.zeros((4, 5), dtype = np.int8)
print(my_arr2d)
print(my_arr2d.shape)

# remplacement des 0 par des 1
my_arr2d = np.ones(my_arr2d.shape, dtype = np.int8)
print(my_arr2d)
print(my_arr2d.shape)

# Creation de la copie du tableau 
my_arr2d_cp = np.copy(my_arr2d)

# Replacement par 99 des valeurs du tableau 
my_arr2d_cp = np.full(my_arr2d_cp.shape, 99)
print(my_arr2d_cp)

# Creation d'un tableau à deux dimensions compris en [0, 1]
my_arr1d_rd = np.random.rand(1000, 3)

# Creation d'un tableau qui represente une simulation de 1000 lancers de 3 dés équilibrés à 6 faces numéroté de 1 à 6
my_arr2d_rd = np.random.randint(1, 7, (1000, 3))
print("Les 5 premières lignes et colonnes : ", my_arr2d_rd[:5])
print("Les 5 derniers lignes et colonnes : ", my_arr2d_rd[-5:])

# Matrice identité du même format que my_arr2d_rd
my_arr_id = np.eye(my_arr2d_rd.shape[0],my_arr2d_rd.shape[1])

print(my_arr_id.shape)
print(my_arr_id[:5])

```

## 4 Manipulation des tableaux

```python
# Construire deux tableaux A et B de format (3, 4) contenant des valeurs aléatoires qui suivent une loi normale centrée réduite

A = np.random.randn(3, 4)
B = np.random.randn(3, 4)
print(A)
print(B)

# Concaténer les tableaux A et B verticalement
C = np.vstack((A, B))
print(C)
print(C.shape)

# Concaténer les tableaux A et B horizontalement
C = np.hstack((A, B))
print(C)
print(C.shape)

# Utiliser la méthode concatenate pour concaténer les tableaux A et B verticalement ou horizontalement.
C = np.concatenate((A, B), axis = 0) # vertical
print(C.shape)
C = np.concatenate((A, B), axis = 1) # horizontal
print(C.shape)

# Utiliser la méthode reshape pour changer le format de tableau A en (6, 2).
A = np.reshape(A, (6,2))
print(A)

# Utiliser la méthode ravel pour vectoriser le tableau B.
B = np.ravel(B)

print(B)
```

# 5 Indexation et sélection
Il est possible d’accéder aux éléments d’un tableau grâce à l’indexation. Il se fait de la même manière avec Numpy qu’avec les listes en Python.

```python
# Création d'une matrice suivant une loi normale de moyenne 4 et de variance racine de 2
np.random.seed(7)
X = np.random.normal(loc=4, scale=np.sqrt(2), size=(10,7)) # mu = 4, signal = racine de 2
print(X)

# Afficher le premier et le dernier élément de X, i.e. x1,1 et x10,7.
print(X[0,0])
print(X[-1, -1])

# Afficher la première et la dernière colonne de X.
print(X[:, 0])
print(X[:, -1])

# Exécuter X[::3, ::3] et commenter.
print(X[::3, ::3]) # prendre une ligne sur trois et une colonne sur trois, en commençant à l'indice 0.

# Créer un tableau booléen mask vérifiant si les valeurs du tableau X sont supérieures à 4.
mask = X>4 
print(mask) 

# Créer un tableau booléen mask1 vérifiant si les valeurs du tableau X sont supérieures à 4 et inférieures à 6.
mask1 = (X>4) & (X <6)  
print(mask1) 

# Exécuter Y = X[mask] et Z = X[mask1].
Y = X[mask]
Z = X[mask1]
```

# 6 Somme lignes et colonnes


```python
# Créer un tableau sumColX contenant la somme de chaque colonne de X.
sumColX = np.sum(X, axis=0)
print(sumColX)

# Créer un tableau sumRawX contenant la somme de chaque ligne de X.
sumRawX = np.sum(X, axis=1)
print(sumRawX)

# Calculer le minimum et maximum de chaque colonne de X.
minColX = np.min(X, axis=0)
maxColX = np.max(X, axis=0)
print(minColX, maxColX)

# Déterminer la position du minimum de chaque colonne de X.
posMinColX = np.argmin(X, axis=0)
print(posMinColX)
```

## 7 Opérations mathématiques et statistiques

```python
# Écrire une fonction qui renvoie un dictionnaire ayant pour clés le numéro de chaque colonne de X et
# pour valeurs un autre dictionnaire qui résume les agrégats suivants: moyenne, variance, écart type,
# médiane, de chaque colonne. 

def statsX(X: np.ndarray) -> dict:
    dicResumeX = {}
    meanColX = np.mean(X, axis=0)
    varColX = np.var(X, axis=0)
    stdColX = np.std(X, axis=0)
    medianColX = np.median(X, axis=0)
    
    for i in range(X.shape[1]):
        name = "column n° " + str(i+1)
        dicResumeX[name] = {
            'moyenne': meanColX[i],
            'variance': varColX[i],
            'std': stdColX[i],
            'median': medianColX[i],
            }
        
    return dicResumeX

dicResumeX = statsX(X)

print(dicResumeX)
```

## 8 Algèbre linéaire

```python
# Calculer Xtop la transposé de la matrice X. Vérifier le format de Xtop.
Xtop = X.T
print(Xtop.shape)

# A l’aide de la méthode dot, calculer le produit matriciel 
P = X.T @ X
print(P.shape)

# Utiliser la méthode np.linalg.eig pour calculer les valeurs et vecteurs propres
eig = np.linalg.eig(P)
print(eig)

# Calculer la décomposition de la valeur singulière de X à l’aide de la méthode np.linalg.svd
U, s, Vtop = np.linalg.svd(X,full_matrices=False)
print(U.shape)
print(s.shape)
print(Vtop.shape)

# On vérifie que les matrices U et V sont des matrices unitaires,
np.set_printoptions(precision=2)
print(U.T.dot(U), sep=2 * '\n')
print(Vtop.T.dot(Vtop), sep=2 * '\n')
```
