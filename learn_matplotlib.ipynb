# Matplotlib

## 1 Méthode plot

```python
# Importer Numpy et le module pyplot de Matplotlib et lui associer l’alias plt.
import numpy as np
import matplotlib.pyplot as plt
# affichage des figures au sein du notebook
%matplotlib inline 

# Déclarer un tableau unidimensionnel x contenant des réels entre 0 et 3π de taille 50.
x = np.linspace(0, 3*np.pi, 50)

# Créer deux tableaux y = sin(x) et z = cos(x).
y = np.sin(x)
z = np.cos(x)

# Tracer y, z en fonction de x avec legende, couleur, marker .
plt.plot(x, y, label=r"$y = \sin(x)$", color='red', marker='o', linestyle='-')
plt.plot(x, z, label=r"$z = \cos(x)$", color='blue', marker='+', linestyle='solid')

plt.title("Représentation de sin(x) et cos(x)") # titre
plt.xlabel("x") # label pour y
plt.ylabel("Valeur")
plt.legend() # Legend
plt.grid() # Pour la grille

plt.show()
```

## 2 Méthode scatter

```python
from sklearn.datasets import load_iris # importation du dataset

# Déclarer une variable iris = load_iris(). Afficher iris, il s’agit d’un dictionnaire.
iris = load_iris()

X = iris.data # contient la longueur, largeur de la petale, et longueur et largeur de sepale
y = iris.target # valeur à predict
names = list(iris.target_names) # type de fleur

# print(X.shape)
# print(y.shape)
# print(names)

# Visualiser les données
# plt.scatter(X[:, 0], X[:, 1], c=y) # longueur, largeur de la petale
plt.scatter(X[:, 2], X[:, -1], c= y) # longueur, largeur de sepale + Colorer les points suivant la nature leurs classes.

plt.xlabel("Longueur du sépale")
plt.ylabel("Largeur du sépale")
plt.title("Nuage de points des fleurs d'Iris") 
plt.show()

# visualisation en 3d
ax = plt.axes(projection='3d') # ax est un objet sur lequel on peut travailler
ax.scatter(X[:, 0], X[:, 1], X[:, 2], c=y)

ax.set_xlabel("Longueur du sépale")
ax.set_ylabel("Largeur du sépale")
ax.set_zlabel("Longueur du pétale")
plt.show()
```

## 3 Visualisation des fonctions mathématiques

$$
f(u,v)=\frac{u^2-v^2}{1+u^2+v^2}
$$

```python
from mpl_toolkits.mplot3d import Axes3D

# Créer une grille de valeurs (U, V) de l’intervalle [−2, 2] (indication utiliser la méthode np.meshgrid de Numpy).
u = np.linspace(-2, 2, 50)
v = np.linspace(-2, 2, 50)

U, V = np.meshgrid(u, v)

print(U.shape)
print(V.shape)

# Déclarer une variable W contenant les images de la grille par la fonction f(U, V).

W = (U**2 - V**2) / (1 + U**2 + V**2)
print(W.shape)

# affichage en 3D
fig= plt.figure()
#ax = Axes3D(fig)
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(U, V, W) 

plt.show()

# Utiliser un code de couleur.
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
surface = ax.plot_surface(U, V, W, cmap=plt.cm.viridis)
fig.colorbar(surface)
ax.view_init(20, 60)

plt.show()
```

## 5 Histogrammes

```python
# A l’aide de la méthode hist de pyplot, tracer les histogrammes 1D de distribution des 4 variables du dataset iris.
fig, axes = plt.subplots(2, 2)
axes[0, 0].hist(X[:, 0])
axes[0, 0].set_title("longueur de sépale")

axes[0, 1].hist(X[:, 1])
axes[0, 1].set_title("largeur de sépale")

axes[1, 0].hist(X[:, 2])
axes[1, 0].set_title("longueur de pétale")

axes[1, 1].hist(X[:, -1])
axes[1, 1].set_title("largeur de pétale")

fig.suptitle("Distribution des variables du dataset Iris")
plt.tight_layout() # ajustement automatiquement l'espace entre les differents graphiques de la figure

plt.show()

# A l’aide de la méthode hist2d de pyplot, tracer les histogrammes 2D de distribution des deux premières variables du dataset iris.
plt.hist2d(X[:, 0], X[:, 1]) 
plt.xlabel("Longueur du sépale")
plt.ylabel("Largeur du sépale")
plt.title("Distribution 2D des deux premières variables")

plt.colorbar(label="Nombre de fleurs")
plt.show()
```
